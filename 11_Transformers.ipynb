```python
# -------------------------------------------------------------------------
# 1. High-Level API: Using Pre-built Pipelines
# -------------------------------------------------------------------------
print("=== 1. Sentiment Analysis Pipeline ===")
    
# Initialize a pipeline (downloads a pre-trained model automatically)
classifier = pipeline("sentiment-analysis")
    
sample_texts = [
    "Transformers have completely revolutionized natural language processing!",
    "The model took too long to train and ran out of memory."
]
    
results = classifier(sample_texts)
for text, result in zip(sample_texts, results):
    print(f"Text: '{text}'")
    print(f"Result: {result['label']} (Score: {result['score']:.4f})\n")
```

```python
print("=== 2. Text Generation (GPT-2) ===")
    
generator = pipeline("text-generation", model="gpt2")
prompt = "Data science with Transformers is powerful because"
    
generated = generator(prompt, max_length=50, num_return_sequences=1)
print(f"Prompt: {prompt}")
print(f"Generated: {generated[0]['generated_text']}\n")
```

```python
print("=== 3. Explicit Tokenization & Model Forward Pass ===")
    
model_name = "bert-base-uncased"
    
# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModel.from_pretrained(model_name)
    
text = "Attention is all you need."
    
# Step A: Convert raw text into input tensors
inputs = tokenizer(text, return_tensors="pt")
    
print(f"Input Tokens: {tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])}")
print(f"Input IDs Shape: {inputs['input_ids'].shape}")
    
# Step B: Pass tensors through the Transformer model
with torch.no_grad():
    outputs = model(**inputs)
    
# Extract the last hidden states (token embeddings)
last_hidden_state = outputs.last_hidden_state
print(f"Output Hidden State Shape: {last_hidden_state.shape}")
print("(Batch Size: 1, Sequence Length: 8, Hidden Dimension: 768)\n")
```
